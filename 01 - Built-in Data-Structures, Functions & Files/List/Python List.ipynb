# Python List

```python
mylist = ["apple", "banana", "cherry"]
print(mylist)
```

```python
# Allowing Duplicates
mylist = ["apple", "banana", "cherry", "apple", "cherry"]
print(mylist)
```

```python
# List Length
mylist = ["apple", "banana", "cherry"]
print(len(mylist))
```

```python
# List Items - Data Types
list1 = ["apple", "banana", "cherry"]
list2 = [1, 5, 7, 9, 3]
list3 = [True, False, False]
```

```python
# Type
mylist = ["apple", "banana", "cherry"]
print(type(mylist))
```

```python
# List Constructor
mylist = list(("apple", "banana", "cherry")) 
print(mylist)
print(type(mylist))
```

# Python - Access List Items

```python
# Access List Items
thislist = ["apple", "banana", "cherry"]
print(thislist[1])
```
